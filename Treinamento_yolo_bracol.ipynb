```python
#Estamos usando o dataset abaixo, possue imagens de folhas saudáveis e doentes
#https://www.kaggle.com/datasets/noamaanabdulazeem/jmuben-coffee-dataset?resource=download-directory
# Instala o Ultralytics YOLO
!pip install ultralytics

import torch
import ultralytics

# Verifica o ambiente
ultralytics.checks()
print("GPU Disponível:", torch.cuda.is_available())
```

```python
# Descompacta o dataset com as imagens de café (é preciso subir o dataset toda vez que rodamos)
!unzip -q bracol-dataset.zip -d /content/bracol-dataset
```

```python
path: /content/bracol-dataset        # Caminho raiz no Colab
train: train/images                  # Caminho para as imagens de treino
val: valid/images                    # Caminho para as imagens de validação

nc: 4                                # Número de classes de doenças do café
names: ['ferrugem', 'bicho-mineiro', 'phoma', 'cercospora']
```

```python
yaml_conteudo = """
path: /content/dataset_yolo
train: train/images
val: val/images

nc: 2
names:
  0: saudavel
  1: doente
"""

with open('/content/dataset_yolo/data.yaml', 'w') as f:
    f.write(yaml_conteudo.strip())

print("Arquivo data.yaml criado com sucesso!")
```

```python
from ultralytics import YOLO

# Carrega o modelo pré-treinado
model = YOLO("yolov8n.pt")

# Inicia o treinamento
results = model.train(
    data="/content/dataset_yolo/data.yaml",
    epochs=50,       # Quantidade de épocas (ajuste conforme necessário)
    imgsz=640,       # Tamanho da imagem
    batch=16         # Tamanho do lote
)
```
